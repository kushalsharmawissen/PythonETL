```python
Import pandas, random, datetime, csv, and helper libraries to read, generate, and write CSV data.</VSCode.Cell>
<VSCode.Cell language="python">import csv
import random
from datetime import datetime, timedelta
from decimal import Decimal
from pathlib import Path

csv_path = Path("sample_transactions.csv")
</VSCode.Cell>
<VSCode.Cell language="markdown"># Load Existing Sample CSV
Read the current CSV, preserve the header row, and inspect the existing row count before appending.</VSCode.Cell>
<VSCode.Cell language="python">with csv_path.open(newline="", encoding="utf-8") as f:
    reader = list(csv.reader(f))
header = reader[0]
existing_rows = reader[1:]
print(f"Existing rows: {len(existing_rows)}")
print(f"Header columns: {len(header)}")
</VSCode.Cell>
<VSCode.Cell language="markdown"># Define Synthetic Transaction Schema
Define fields and generation rules for transaction IDs, dates, amounts, customer/merchant details, and status values.</VSCode.Cell>
<VSCode.Cell language="python">random.seed(42)

channels = ["Mobile App", "POS", "ATM", "Branch", "Web", "Call Center"]
currencies = ["INR", "USD", "EUR", "GBP", "JPY", "AUD"]
transaction_types = ["debit", "credit"]
categories = ["SHOPPING", "FOOD", "UTILITIES", "TRAVEL", "HEALTH", "ENTERTAINMENT"]
sub_categories = {
    "SHOPPING": ["ELECTRONICS", "FASHION", "GROCERY"],
    "FOOD": ["RESTAURANT", "CAFE", "GROCERY"],
    "UTILITIES": ["ELECTRICITY", "WATER", "INTERNET"],
    "TRAVEL": ["FLIGHT", "HOTEL", "RAIL"],
    "HEALTH": ["PHARMACY", "CLINIC", "FITNESS"],
    "ENTERTAINMENT": ["MOVIES", "MUSIC", "EVENTS"],
}
merchants = ["Zomato", "Dominos", "PowerCo", "Myntra", "Air India", "HealthPlus", "BookStore", "CinemaWorld", "PayFast", "RideShare"]
cities = ["Bangalore", "Mumbai", "Delhi", "Chennai", "Pune", "Kolkata", "Hyderabad", "Ahmedabad"]
countries = ["IN", "US", "GB", "JP", "AU"]
payment_methods = ["Card", "NEFT", "IMPS", "UPI", "Cash"]
segments = ["RETAIL", "CORPORATE", "WEALTH", "HNI"]
timezones = ["Asia/Kolkata", "US/Eastern", "Europe/London", "Asia/Tokyo", "Australia/Sydney"]
status_choices = ["SUCCESS", "FAILED", "PENDING"]
recurring_options = ["yes", "no"]
account_types = ["SAVINGS", "CURRENT", "LOAN", "CREDIT"]
</VSCode.Cell>
<VSCode.Cell language="markdown"># Generate 10,000 Additional Transactions
Construct synthetic rows with realistic date, currency, and merchant values while avoiding duplicate transaction IDs.
</VSCode.Cell>
<VSCode.Cell language="python">start_index = len(existing_rows) + 1
new_rows = []
for i in range(10000):
    idx = start_index + i
    txn_id = f"TXN{idx:05d}"
    txn_date = (datetime(2024, 1, 1) + timedelta(days=random.randint(0, 180))).strftime("%Y-%m-%d")
    txn_time = f"{random.randint(0, 23):02d}:{random.randint(0, 59):02d}:{random.randint(0, 59):02d}"
    channel = random.choice(channels)
    currency = random.choice(currencies)
    amount = Decimal(random.randint(1000, 2000000)) / Decimal("100")
    ttype = random.choice(transaction_types)
    category = random.choice(categories)
    sub_category = random.choice(sub_categories[category])
    merchant = random.choice(merchants)
    city = random.choice(cities)
    country = random.choice(countries)
    payment = random.choice(payment_methods)
    is_rec = random.choice(recurring_options)
    status = random.choices(status_choices, weights=[0.8, 0.1, 0.1])[0]
    exchange = Decimal(random.uniform(0.5, 90)).quantize(Decimal("0.0001")) if currency != "INR" else Decimal("1.0000")
    notes = "Sample generated transaction"
    account_id = f"ACC{random.randint(100, 999):03d}"
    account_type = random.choice(account_types)
    customer_id = f"CUST{random.randint(10, 999):03d}"
    customer_name = f"Customer{random.randint(100, 999)}"
    customer_email = f"user{random.randint(1000, 9999)}@example.com"
    customer_phone = f"9{random.randint(100000000, 999999999)}"
    segment = random.choice(segments)
    timezone = random.choice(timezones)
    rewards = random.randint(0, 15000)
    new_rows.append([
        txn_id,
        txn_date,
        txn_time,
        channel,
        currency,
        f"{amount:.2f}",
        ttype,
        category,
        sub_category,
        merchant,
        city,
        country,
        payment,
        is_rec,
        status,
        f"{exchange:.4f}",
        notes,
        account_id,
        account_type,
        customer_id,
        customer_name,
        customer_email,
        customer_phone,
        segment,
        timezone,
        rewards,
    ])
print(f"Generated {len(new_rows)} new rows")
</VSCode.Cell>
<VSCode.Cell language="markdown"># Append New Rows to CSV
Write only the generated rows to the existing CSV and preserve the original header row.
</VSCode.Cell>
<VSCode.Cell language="python">with csv_path.open("a", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(new_rows)
print(f"Appended {len(new_rows)} rows to {csv_path}")
</VSCode.Cell>
<VSCode.Cell language="markdown"># Verify Updated Row Count
Reload the CSV and confirm the dataset now contains the original rows plus the 10,000 new rows.
</VSCode.Cell>
<VSCode.Cell language="python">with csv_path.open(newline="", encoding="utf-8") as f:
    total_rows = sum(1 for _ in csv.reader(f))
print(f"Total rows including header: {total_rows}")
print(f"Total data rows: {total_rows - 1}")
```
